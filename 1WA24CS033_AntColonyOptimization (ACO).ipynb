```python

import random
import math

# City coordinates
cities = [
    (0, 0),
    (1, 5),
    (5, 2),
    (6, 6),
    (8, 3)
]

N = len(cities)

# Parameters
ANTS = 10
ITERATIONS = 50
ALPHA = 1
BETA = 2
RHO = 0.5

# Calculate distance
def distance(i, j):
    x1, y1 = cities[i]
    x2, y2 = cities[j]
    return math.sqrt((x2-x1)**2 + (y2-y1)**2)

# Initialize pheromone
pheromone = [[1.0 for _ in range(N)]
             for _ in range(N)]

best_route = None
best_distance = float('inf')

for iteration in range(ITERATIONS):

    all_routes = []

    for ant in range(ANTS):

        route = [random.randrange(N)]

        # Construct route
        while len(route) < N:

            current = route[-1]
            choices = [i for i in range(N)
                       if i not in route]

            probabilities = []

            for city in choices:
                tau = pheromone[current][city] ** ALPHA
                eta = (1 / distance(current, city)) ** BETA
                probabilities.append(tau * eta)

            total = sum(probabilities)
            probabilities = [p / total for p in probabilities]

            next_city = random.choices(
                choices, weights=probabilities
            )[0]

            route.append(next_city)

        # Calculate total route distance
        total_distance = sum(
            distance(route[i], route[(i+1) % N])
            for i in range(N)
        )

        all_routes.append((route, total_distance))

        if total_distance < best_distance:
            best_distance = total_distance
            best_route = route.copy()

    # Evaporate pheromone
    for i in range(N):
        for j in range(N):
            pheromone[i][j] *= (1 - RHO)

    # Deposit pheromone
    for route, length in all_routes:
        deposit = 1 / length

        for i in range(N):
            a = route[i]
            b = route[(i+1) % N]

            pheromone[a][b] += deposit
            pheromone[b][a] += deposit

# Output
print("Best Route:", best_route + [best_route[0]])
print("Shortest Distance:", best_distance)
```
