```python

import numpy as np

def fitness(x):
    return sum(x**2)

n = 10
dim = 2
iterations = 50

wolves = np.random.uniform(-10, 10, (n, dim))

for t in range(iterations):
    scores = np.array([fitness(w) for w in wolves])
    idx = np.argsort(scores)

    alpha = wolves[idx[0]]
    beta = wolves[idx[1]]
    delta = wolves[idx[2]]

    a = 2 - 2 * t / iterations

    for i in range(n):
        A1 = 2*a*np.random.rand(dim) - a
        C1 = 2*np.random.rand(dim)
        D1 = abs(C1*alpha - wolves[i])
        X1 = alpha - A1*D1

        A2 = 2*a*np.random.rand(dim) - a
        C2 = 2*np.random.rand(dim)
        D2 = abs(C2*beta - wolves[i])
        X2 = beta - A2*D2

        A3 = 2*a*np.random.rand(dim) - a
        C3 = 2*np.random.rand(dim)
        D3 = abs(C3*delta - wolves[i])
        X3 = delta - A3*D3

        wolves[i] = (X1 + X2 + X3) / 3
        wolves[i] = np.clip(wolves[i], -10, 10)

best = wolves[np.argmin([fitness(w) for w in wolves])]

print("Best Position:", best)
print("Minimum Fitness:", fitness(best))
```
