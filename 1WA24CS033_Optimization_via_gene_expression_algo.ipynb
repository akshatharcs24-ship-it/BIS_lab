```python
import random

# Distance matrix between 4 locations
distance = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

# Parameters
POP_SIZE = 10
GENERATIONS = 50
MUTATION_RATE = 0.1

# Generate random population
def create_population():
    return [
        random.sample(range(4), 4)
        for _ in range(POP_SIZE)
    ]

# Gene expression: Decode chromosome
def gene_expression(chromosome):
    return chromosome

# Calculate fitness
def fitness(chromosome):
    route = gene_expression(chromosome)
    total = 0

    for i in range(len(route) - 1):
        total += distance[route[i]][route[i + 1]]

    total += distance[route[-1]][route[0]]

    return 1 / total

# Selection
def selection(population):
    return max(population, key=fitness)

# Crossover
def crossover(parent1, parent2):
    point = random.randint(1, 3)
    child = parent1[:point]

    for gene in parent2:
        if gene not in child:
            child.append(gene)

    return child

# Mutation
def mutation(child):
    if random.random() < MUTATION_RATE:
        i, j = random.sample(range(4), 2)
        child[i], child[j] = child[j], child[i]

    return child

# Main GEA
population = create_population()

for generation in range(GENERATIONS):

    new_population = []

    for _ in range(POP_SIZE):
        parent1 = selection(population)
        parent2 = random.choice(population)

        child = crossover(parent1, parent2)
        child = mutation(child)

        new_population.append(child)

    population = new_population

# Best solution
best_route = max(population, key=fitness)

print("Best Delivery Route:", best_route)
print("Minimum Distance:",
      1 / fitness(best_route))
```
