```python

import random

# Fitness function
def fitness(x):
    return x ** 2

# Parameters
NUM_PARTICLES = 10
ITERATIONS = 50
W = 0.5
C1 = 1.5
C2 = 1.5

# Initialize particles
positions = [random.uniform(-10, 10)
             for _ in range(NUM_PARTICLES)]

velocities = [random.uniform(-1, 1)
              for _ in range(NUM_PARTICLES)]

pbest = positions.copy()

gbest = max(pbest, key=fitness)

# PSO iterations
for _ in range(ITERATIONS):

    for i in range(NUM_PARTICLES):

        r1 = random.random()
        r2 = random.random()

        # Update velocity
        velocities[i] = (
            W * velocities[i]
            + C1 * r1 * (pbest[i] - positions[i])
            + C2 * r2 * (gbest - positions[i])
        )

        # Update position
        positions[i] += velocities[i]

        # Boundary checking
        positions[i] = max(-10, min(10, positions[i]))

        # Update personal best
        if fitness(positions[i]) > fitness(pbest[i]):
            pbest[i] = positions[i]

    # Update global best
    current_best = max(pbest, key=fitness)

    if fitness(current_best) > fitness(gbest):
        gbest = current_best

# Output
print("Best Solution:", gbest)
print("Maximum Fitness:", fitness(gbest))
```
