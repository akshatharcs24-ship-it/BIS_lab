```python

import random
from concurrent.futures import ThreadPoolExecutor

# Objective function
def fitness(x):
    return x * x

# Initialize parameters
N = 10
MAX_ITER = 100
MIN = -10
MAX = 10
alpha = 0.5

# Initialize cell population
position = [random.uniform(MIN, MAX) for _ in range(N)]

print("Parallel Cellular Algorithm")
print("Initial Population:")
print(position)

# Function to update each cell
def update_cell(i):
    left = (i - 1 + N) % N
    right = (i + 1) % N

    # Find best neighboring cell
    neighbors = [position[left], position[i], position[right]]
    best = min(neighbors, key=fitness)

    # Update cell position
    new_position = position[i] + alpha * (best - position[i])

    # Keep position within search range
    new_position = max(MIN, min(MAX, new_position))

    return new_position

# Parallel execution
with ThreadPoolExecutor(max_workers=N) as executor:

    for iteration in range(MAX_ITER):

        # Update all cells simultaneously
        new_position = list(
            executor.map(update_cell, range(N))
        )

        position = new_position

        if (iteration + 1) % 10 == 0:
            best = min(position, key=fitness)
            print("Iteration:", iteration + 1,
                  "Best X:", round(best, 6),
                  "Fitness:", round(fitness(best), 8))

# Find final best solution
best_solution = min(position, key=fitness)

print("\nFinal Result")
print("Best Solution (x):", best_solution)
print("Minimum Fitness:", fitness(best_solution))
```
