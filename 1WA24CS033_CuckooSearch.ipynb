```python

import random

def fitness(x):
    return x * x

n = 10
iterations = 50
nests = [random.uniform(-10, 10) for i in range(n)]

best = min(nests, key=fitness)

for i in range(iterations):
    for j in range(n):
        new = nests[j] + random.uniform(-1, 1)

        if fitness(new) < fitness(nests[j]):
            nests[j] = new

        if random.random() < 0.25:
            nests[j] = random.uniform(-10, 10)

    best = min(nests, key=fitness)

print("Best solution:", best)
print("Minimum fitness:", fitness(best))
```
